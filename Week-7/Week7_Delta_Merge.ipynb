```python
df = spark.table("sample_superstore")

display(df)
```

```python
from pyspark.sql.functions import col, when, count

# Count missing values in each column
missing_values = df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in df.columns
])

display(missing_values)
```

```python
clean_df = df.dropna()

print("Original Rows:", df.count())
print("Rows After Removing Nulls:", clean_df.count())

display(clean_df)
```

```python
clean_df.select("Order ID", "Customer Name", "Sales").show(10)
```

```python
high_sales = clean_df.filter(clean_df["Sales"] > 500)

display(high_sales)
```

```python
technology = clean_df.filter(clean_df["Category"] == "Technology")

display(technology)
```

```python
# Count rows before removing duplicates
print("Rows before removing duplicates:", clean_df.count())

# Remove duplicate rows
clean_df = clean_df.dropDuplicates()

# Count rows after removing duplicates
print("Rows after removing duplicates:", clean_df.count())

# Display the cleaned data
display(clean_df)
```

```python
from pyspark.sql.functions import col

clean_df = clean_df.withColumn(
    "Unit_Price",
    col("Sales") / col("Quantity")
)
```

```python
clean_df = clean_df.withColumn(
    "total_amount",
    col("Unit_Price") * col("Quantity")
)

display(clean_df)
```

```python
# Replace spaces with underscores
for old_name in clean_df.columns:
    new_name = old_name.replace(" ", "_")
    clean_df = clean_df.withColumnRenamed(old_name, new_name)

display(clean_df)
```

```python
# Convert Spark DataFrame to Pandas
pdf = clean_df.toPandas()

# Save as CSV
pdf.to_csv("Cleaned_Superstore.csv", index=False)

print("Cleaned_Superstore.csv created successfully.")
```
